### Introduction

In this syllabus, you'll learn the CUDA Tile programming model and use cuTile Python to generate Tile IR kernels.

CUDA Tile is an array-oriented parallel programming model. You divide your inputs into local arrays called tiles that are processed concurrently by groups of execution resources called blocks. You write tile kernels that are executed once per block. Array operations within a kernel are parallelized across the block. The framework handles parallelization, synchronization, and data movement behind the scenes. Architecture-specific details are abstracted away: you express your algorithm once in a high-level way, and the system lowers it for the target NVIDIA GPU architecture.

These lessons require an NVIDIA R580-or-newer driver and an Ampere, Ada, or Blackwell GPU; TileIRAs 13.2 does not support Hopper. The shared tutorial image uses CUDA Toolkit 13.2.1 and cuTile Python 1.4.0 with its matching TileIRAs compiler.

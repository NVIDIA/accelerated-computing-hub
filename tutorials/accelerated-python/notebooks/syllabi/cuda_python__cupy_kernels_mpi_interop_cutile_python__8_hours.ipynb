### Introduction

This eight-hour CUDA Python course follows the foundations of the two-day PyHPC course, but replaces its Shallow Water Equations case study with a focused CUDA Tile exercise sequence.

We start with the NumPy and CuPy array model, move data deliberately between host and device, profile asynchronous execution, and author custom CUDA kernels with Numba CUDA. We then distribute work with mpi4py, connect Python and C++, and finish by expressing vector, matrix, transpose, and activation kernels with the higher-level CUDA Tile programming model.

In this tutorial we will cover:
- NumPy fundamentals and the ndarray memory model.
- Accelerating array workflows on the GPU with CuPy, including memory spaces and asynchrony.
- Authoring and profiling custom CUDA kernels with Numba CUDA, Nsight Systems, and Nsight Compute.
- Writing array-oriented GPU kernels with cuTile Python.
- Distributed computing with mpi4py.
- Python/C++ interoperability with ctypes, cffi, nanobind, and CppJIT.

The CUDA Tile lessons require an NVIDIA R580-or-newer driver and an Ampere, Ada, or Blackwell GPU; TileIRAs 13.2 does not support Hopper. Attendees are expected to have general knowledge of Python programming. Some familiarity with NumPy and C/C++ is helpful but not required.

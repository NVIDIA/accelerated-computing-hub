### Introduction

This eight-hour CUDA Python course focuses entirely on CuPy, custom CUDA kernels, and CUDA Tile. It selects the corresponding foundations from the two-day PyHPC course and replaces its longer case study with a focused CUDA Tile exercise sequence.

We start by moving array workflows to CuPy, move data deliberately between host and device, profile asynchronous execution, and author custom CUDA kernels with Numba CUDA. We finish by expressing vector, matrix, transpose, and activation kernels with the higher-level CUDA Tile programming model.

In this tutorial we will cover:
- Accelerating array workflows on the GPU with CuPy, including memory spaces and asynchrony.
- Authoring and profiling custom CUDA kernels with Numba CUDA, Nsight Systems, and Nsight Compute.
- Writing array-oriented GPU kernels with cuTile Python.

The CUDA Tile lessons require an NVIDIA R580-or-newer driver and an Ampere, Ada, or Blackwell GPU; TileIRAs 13.2 does not support Hopper. Attendees are expected to have general knowledge of Python programming. Some familiarity with NumPy is helpful but not required.

### Introduction

This eight-hour CUDA Python course focuses entirely on CuPy, CUDA SIMT kernels, and CUDA Tile. It selects the corresponding foundations from the two-day PyHPC course and replaces its longer case study with a focused CUDA Tile exercise sequence.

We start by moving array workflows to CuPy, moving data deliberately between host and device, and profiling asynchronous execution. We then express vector, matrix multiplication, transpose, and activation kernels with the higher-level CUDA Tile programming model before finishing with custom SIMT kernels in Numba CUDA.

In this tutorial we will cover:
- Accelerating array workflows on the GPU with CuPy, including memory spaces, asynchrony, and Nsight Systems profiling.
- Writing array-oriented GPU kernels with cuTile Python.
- Authoring and profiling custom CUDA SIMT kernels with Numba CUDA and Nsight Compute.

Attendees are expected to have general knowledge of Python programming. Some familiarity with NumPy is helpful but not required.
